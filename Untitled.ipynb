```python
# wms_minimal_jax_units.py
import jax
import jax.numpy as jnp

PI2 = 2.0 * jnp.pi


# -------------------- timebase (seconds) --------------------
def make_timebase(fm_hz: float, fs_hz: float, n_periods: int = 1):
    """
    Sample an integer number of modulation periods at a fixed sample rate.

    Args:
      fm_hz: modulation frequency [Hz]
      fs_hz: sample rate [Hz]
      n_periods: number of periods [unitless]

    Returns:
      t_s: time array [s]
      dt_s: sample spacing [s]
      n_per_period: samples per period [samples]
    """
    dt_s = 1.0 / fs_hz
    n_per_period_f = fs_hz / fm_hz
    n_per_period = int(round(n_per_period_f))

    # For clean lock-in over an integer number of cycles, we want fs/fm to be an integer.
    if abs(n_per_period - n_per_period_f) > 1e-12:
        raise ValueError(
            f"fs_hz/fm_hz must be (very nearly) an integer for this simple setup. "
            f"Got fs/fm = {n_per_period_f}."
        )

    t_s = dt_s * jnp.arange(n_per_period * n_periods)
    return t_s, dt_s, n_per_period


# -------------------- detuning waveform (Hz) --------------------
def detuning_hz_sine(t_s, fm_hz: float, m1: float, hwhm_hz: float):
    """
    Δν(t) [Hz] = (m1 * HWHM) * sin(2π f_m t)
    where m1 is a modulation index in units of HWHM.
    """
    theta = PI2 * fm_hz * t_s
    return (m1 * hwhm_hz) * jnp.sin(theta)


def detuning_hz_1f3f(t_s, fm_hz: float, m1: float, m3: float, phi3_rad: float, hwhm_hz: float):
    """
    Δν(t) [Hz] = HWHM * [ m1*sin(θ) + m3*sin(3θ + φ3) ]
    """
    theta = PI2 * fm_hz * t_s
    return hwhm_hz * (m1 * jnp.sin(theta) + m3 * jnp.sin(3.0 * theta + phi3_rad))


# -------------------- absorption placeholder --------------------
def transmission_placeholder(detuning_hz, peak_abs: float, hwhm_hz: float):
    """
    A(Δν) = peak_abs * 1/(1 + (Δν/HWHM)^2),   I = exp(-A)

    Swap A(Δν) with your real absorbance table later.
    """
    x = detuning_hz / hwhm_hz
    A = peak_abs * (1.0 / (1.0 + x * x))
    return jnp.exp(-A)


# -------------------- lock-in (vectorized over harmonics) --------------------
def lockin_amps(signal, t_s, f_ref_hz: float, harmonics=(1, 2, 3)):
    """
    Complex lock-in coefficient for each harmonic h:
      c_h = (2/N) Σ s(t) exp(-i 2π h f t)
      amp_h = |c_h|

    This is JAX-safe (no Python int casts inside vmap).
    """
    hs = jnp.asarray(harmonics, dtype=jnp.float32)  # [H]
    w = PI2 * f_ref_hz * hs[:, None]                # [H,1]
    ref = jnp.exp(-1j * w * t_s[None, :])           # [H,N]
    c = (2.0 / signal.size) * jnp.sum(signal[None, :] * ref, axis=1)  # [H]
    return jnp.abs(c)


# -------------------- end-to-end sim --------------------
def simulate_wms(
    fm_hz: float = 20_000.0,          # [Hz]
    fs_hz: float = 2_000_000.0,       # [Hz]  <-- requested
    n_periods: int = 1,               # [periods]
    waveform: str = "sine",           # "sine" or "1f3f"
    m1: float = 2.25,                 # [unitless, in HWHM units]
    m3: float = 0.0,                  # [unitless, in HWHM units]
    phi3_rad: float = 0.0,            # [rad]
    peak_abs: float = 0.5,            # [dimensionless]
    hwhm_hz: float = 1.0e9,           # [Hz] placeholder scale
    noise_std_time: float = 1e-3,     # per-sample noise std on transmission [dimensionless]
    harmonics=(2,),                   # harmonics to return
    return_trace: bool = False,
):
    """
    Returns:
      amps: lock-in amplitudes for requested harmonics [dimensionless]
      snrs: amps / sigma_amp [dimensionless]
      slew_rms_hz_per_s, slew_max_hz_per_s: [Hz/s]
      dt_s, n_per_period: sampling info
      (optional) detuning_hz, transmission arrays
    """
    t_s, dt_s, n_per_period = make_timebase(fm_hz, fs_hz, n_periods=n_periods)

    if waveform == "sine":
        det_hz = detuning_hz_sine(t_s, fm_hz=fm_hz, m1=m1, hwhm_hz=hwhm_hz)
    elif waveform == "1f3f":
        det_hz = detuning_hz_1f3f(t_s, fm_hz=fm_hz, m1=m1, m3=m3, phi3_rad=phi3_rad, hwhm_hz=hwhm_hz)
    else:
        raise ValueError(f"Unknown waveform='{waveform}' (use 'sine' or '1f3f').")

    I = transmission_placeholder(det_hz, peak_abs=peak_abs, hwhm_hz=hwhm_hz)

    amps_all = lockin_amps(I, t_s, f_ref_hz=fm_hz, harmonics=harmonics)

    # slope metrics in physical units [Hz/s]
    ddt = jnp.diff(det_hz) / dt_s
    slew_rms = jnp.sqrt(jnp.mean(ddt * ddt))
    slew_max = jnp.max(jnp.abs(ddt))

    # simple white-noise lock-in amplitude uncertainty
    N = t_s.size
    sigma_amp = 2.0 * noise_std_time / jnp.sqrt(N)
    snrs = amps_all / sigma_amp

    if return_trace:
        return amps_all, snrs, slew_rms, slew_max, dt_s, n_per_period, det_hz, I
    return amps_all, snrs, slew_rms, slew_max, dt_s, n_per_period


# -------------------- tiny demo --------------------
if __name__ == "__main__":
    # With fm=20 kHz and fs=2 MHz => n_per_period = 100 samples/period
    (amp2_sin,), (snr2_sin,), rms_sin, max_sin, dt_s, npp = simulate_wms(
        waveform="sine", m1=2.25, harmonics=(2,)
    )

    (amp2_13,), (snr2_13,), rms_13, max_13, *_ = simulate_wms(
        waveform="1f3f", m1=4.375, m3=1.88125, phi3_rad=jnp.pi, harmonics=(2,)
    )

    fs_hz = 1.0 / float(dt_s)
    print(f"fm_hz = 20 kHz, fs_hz = {fs_hz/1e6:.3f} MHz, n_per_period = {npp}")
    print("2f amp (sine):   ", float(amp2_sin))
    print("2f amp (1f+3f):  ", float(amp2_13))
    print("2f gain:         ", float(amp2_13 / amp2_sin))
    print("2f SNR (sine):   ", float(snr2_sin))
    print("2f SNR (1f+3f):  ", float(snr2_13))
    print("slew_rms [Hz/s] (sine):  ", float(rms_sin), "slew_max [Hz/s]:", float(max_sin))
    print("slew_rms [Hz/s] (1f+3f): ", float(rms_13),  "slew_max [Hz/s]:", float(max_13))
```
